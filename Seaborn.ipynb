```python
# ScatterPlot()

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

student_df = pd.DataFrame({
    "Study_hours": [2, 3, 5, 1, 4, 6, 2, 5,  7, 3, 2, 2, 2, 3,],
    "Marks": [80, 75, 90, 60, 85, 95, 70, 85, 95, 92, 85, 86, 80, 78]
})
sns.scatterplot(
    data = student_df,
    x= "Study_hours",
    y = "Marks"
)

plt.title("Study hours vs Marks")
plt.show()
```

```python
# Bar Plot()

items_df = pd.DataFrame({
    "Items" : ["Laptop", "Mobile", "EarPhones", "Mouse", "Pods"],
    "Prices" : [980,700,480,570,460]
})

sns.barplot(
    data = items_df,
    x = "Items",
    y = "Prices"
)
plt.title("Items vs Prices")
plt.show()
```

```python
# Count Plot()

student_df = pd.DataFrame({
    "Department" : ["AIML", "ECE", "CSE", "DS", "EEE", "AIML", "ECE"],
    "Students" : [200, 150, 320, 120, 80, 78, 70]
})

sns.countplot(
    data = student_df,
    x = "Department"
)
plt.title("Department vs Students")
plt.show()
```

```python
# Histogram

sns.histplot(
    data = student_df,
    x = "Students",
    bins = 5,
    kde = True
)
plt.title("Students")
plt.show()
```

```python
# Box plot()

sns.boxplot(
    data = student_df,
    x = "Department",
    y = "Students"
)

plt.title("Department vs Students")
plt.show()
```

```python
# Correlation Heatmap

student_df = pd.DataFrame({
    "Study_hours": [2, 3, 5, 1, 4, 6, 2, 5,  7, 3, 2, 2, 2, 3,],
    "Marks": [80, 75, 90, 60, 85, 95, 70, 85, 95, 92, 85, 86, 80, 78]
})
Values = [
    "Study_hours",
    "Marks"
]

correlation_matrix = student_df[
    Values
].corr()

sns.heatmap(correlation_matrix, annot = True)

plt.title("Correlation Heatmap")
plt.show()
```
