#Supervised Learning
It is a type of Machine Learning where the model learns from labeled data and predicts outputs for new data.

Exapmles -
  1. House Prediction
  2. Spam Detection
  3. Marks Prediction

# Types of Supervised Learning
#Regression -
  It is a machine learning technique used to predict a continuous numeric value based on input features.

Examples -
  1. House Price
  2. Salary
  3. Sales Amount

#Classification -
  It is a type of supervised learning in which the target variable (output) is a category or label.

Examples -
  1. Spam/Not Spam
  2. Pass/ Fail
  3. Disease/No Disease
